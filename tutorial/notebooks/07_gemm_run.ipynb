```python
import sys
from pathlib import Path

repo_root = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(repo_root))

from tutorial._infra import torch_frontend as torch_nb
from tutorial._infra import nbtools

gemm_mlir = torch_nb.ARTIFACTS_DIR / "gemm_cinm10.mlir"
gemm_ll = nbtools.mlir_translate(gemm_mlir)
print(gemm_ll.read_text()[:2000])
```

```python
import tutorial._infra.target_helpers as th

llvm_ll = torch_nb.ARTIFACTS_DIR  / "gemm_cinm10.ll"
driver_c = torch_nb.DRIVERS_DIR / "gemm_driver.c"
output   = torch_nb.BINARY_OUTPUT_DIR / "program.out"

try:
    print("=== COMPILE ===")
    comp_log = th.compile_binary(driver_c, llvm_ll, output)
    print(comp_log)

    print("=== RUN ===")
    run_log = th.run_binary(output)
    print(run_log)
except Exception as e:
    print("ERROR:", e)
```
